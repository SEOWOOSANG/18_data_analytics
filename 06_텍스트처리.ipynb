```python
# 문자열 메소드

import pandas as pd
import numpy as np

ser = pd.Series(
    ["Apple_사과", "Banana_바나나", "Cherry_체리", np.nan],
    index=["First ", " Second", " Third", "Fourth"])

print(ser)
print()

ser2 = pd.Series(
    ["Apple_사과", "Banana_바나나", "Cherry_체리", np.nan],
    index=["First ", " Second", " Third", "Fourth"], dtype="string")
print(ser2)
```
