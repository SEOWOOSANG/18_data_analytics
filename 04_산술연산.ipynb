```python
# 시리즈연산
# Series객체 + 연산자(+, -, *, /) + 숫자(스칼라)

import pandas as pd

dict_data = {
    "국어": 100,
    "영어": 80,
    "수학": 90
}

student1 = pd.Series(dict_data)
student1
```

```python
persentage = student1/100

print(persentage)
print(type(persentage))
```

```python
# 시리즈 vs 시리즈
# Series + 연산자(+, -, *, /) + Series2

student2 = pd.Series({"수학": 80,"국어": 90, "영어": 80})

print(student1)
print()
print(student2)
```

```python
addition = student1 + student2
addition
```

```python
subraction = student1 - student2
subraction
```

```python
multplication = student1 * student2
multplication
```

```python
division = student1 / student2
division
```

```python
result = pd.DataFrame([addition, subraction, multplication, division],
                      index=["덧셈", "뺄셈", "곱셈", "나눗셈"])

result
```

```python
# NaN 값이 있는 시리즈 연산

import pandas as pd
import numpy as np

student1 = pd.Series({"국어":np.nan, "영어":80, "수학":90})
student2 = pd.Series({"수학":80, "국어": 30})

print(student1)
print()
print(student2)
```

```python
addition = student1 + student2
subraction = student1 - student2
multplication = student1 * student2
division = student1 / student2

print(addition)
print(subraction)
print(multplication)
print(division)
```

```python
result = pd.DataFrame([addition, subraction, multplication, division],
                      index=["덧셈", "뺄셈", "곱셈", "나눗셈"])

result
```
