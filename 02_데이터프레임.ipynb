```python
# 행, 열의 위치 바꾸기: DataFrame 객체.transpose() 또는 DataFrame 객체.T

exam_data = {
    "이름": ["서준", "우현", "인아"],
    "수학": [90,80,70],
    "영어": [98,89,95],
    "음악": [85,95,100],
    "체육": [100,90,90]
}

import pandas as pd

df = pd.DataFrame(exam_data)
df
```
