```python
# 시리즈 원소 선택
import pandas as pd

tup_data = ("영인", "2010-05-01", "여", True)
sr = pd.Series(tup_data, index=["이름", "생년월일", "성별", "학생여부"])
print(sr)
```

```python
# 시리즈 원소 선택

print(sr["이름"])
print(sr["성별"])
```

```python
print(sr[["생년월일", "성별"]])
```

```python
print(sr[1 : 2])
print("\n")
print(sr["생년월일" : "성별"])
```

```python
print(sr["생년월일" : "학생여부"])
print("\n")
print(sr[["생년월일", "학생여부"]])
print("\n")
print(sr.iloc[1:3])
```
