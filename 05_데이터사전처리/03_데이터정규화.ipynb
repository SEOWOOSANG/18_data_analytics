```python
import pandas as pd

df = pd.read_csv("./data/auto-mpg.csv", header=None)

df.columns = [
    "mpg", "cylinders", "displacement", "horsepower", "weight", 
    "acceleration", "model_year", "origin", "name"]

df.head(3)
```

```python
mpg_to_kpl = 1.60934 / 3.78541

df["kpl"] = df["mpg"] * mpg_to_kpl
df.head(3)
```

```python
df["kpl"] = df["kpl"].round(2)

df.head(3)
```
